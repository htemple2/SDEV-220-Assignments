```python
import os
from datetime import datetime

date_string = datetime.now().strftime("%B %d, %Y")

with open("today.txt", "w") as file:
    file.write(date_string)

current_directory = os.getcwd()
print("The file 'today.txt' should be in:", current_directory)

with open("today.txt", "r") as file:
    today_string = file.read()
print("Today's date is:", today_string)

parsed_date = datetime.strptime(today_string, "%B %d, %Y")
print("Parsed datetime:", parsed_date)
```

```python
import multiprocessing
import time
import random
from datetime import datetime

def print_wait(process_number):
    sleep_time = random.random()
    time.sleep(sleep_time)
    current_time = datetime.now().strftime("%H:%M:%S.%f")[:-3]
    print(f"Process {process_number} slept for {sleep_time:.4f} seconds and finished at: {current_time}")

if __name__ == '__main__':
    processes = []
    
    for i in range(3):
        process = multiprocessing.Process(target=print_wait, args=(i+1,))
        processes.append(process)
        process.start()
    for process in processes:
        process.join()
    print(print_wait(1))
    print(print_wait(2))
    print(print_wait(3))
    print("All processes executed.")
    
```
